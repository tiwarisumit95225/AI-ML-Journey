### Filtering Data

```python
bios=pd.read_csv('./data/bios.csv')
```

```python
bios
```

```python
bios.info()
```

```python
bios.loc[bios['height_cm']>215,["name","height_cm"]]
```

```python
bios[(bios['height_cm']>180)&(bios["NOC"]=="India")][["name","height_cm"]]
```

```python
bios[bios["name"].str.contains("Neeraj",case=False)]
```

```python
bios[bios['born_country'].isin(["USA","IND"]) & (bios['name'].str.startswith("Sumit"))]
```

```python
bios.query('born_country=="India" and name=="Sumit"')
```

### Adding/Removing Columns from dataframe

```python

coffee['Price']=5.99
```

```python
coffee
```

```python
import numpy as np

coffee['new_price']=np.where(coffee['Coffee Type']=='Espresso',3.99,6.33)
```

```python
coffee
```

```python
coffee.drop(columns="Price" ,inplace=True)
```

```python
coffee
```

```python
coffee
```

```python
coffee
```

```python
coffee['revenue']=coffee['new_price']*coffee["Units Sold"]
```

```python
coffee
```

```python
coffee=coffee.rename(columns={'new_price':"price"})
```

```python
coffee.to_csv('./data/coffee.csv')
```

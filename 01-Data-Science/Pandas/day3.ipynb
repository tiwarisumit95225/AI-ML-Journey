### Adding/Removing Column

```python
import pandas as pd
```

```python
bios=pd.read_csv('./data/bios.csv')
bios['height_catogery']=bios['height_cm'].apply( lambda x: 'Short' if x < 165 else ('Average' if x<185 else 'Tall') )
```

```python
bios.head()
```

```python
def categorize_athelete(row):
    if row['height_cm']< 175 and row['weight_kg'] <70:
        return 'Lightweight'
    elif  row['height_cm']< 185 or row['weight_kg'] <80:
        return 'Middleweight'
    else:
        return 'Heavyweight'
    
bios['Category']=bios.apply(categorize_athelete,axis=1)
```

```python
bios.head()
```

## Mearging and Concatenating data

```python
nocs=pd.read_csv('./data/noc_regions.csv')
```

```python
bios_new=pd.merge(bios,nocs,left_on='born_country',right_on='NOC',how='left')
```

```python
bios_new.rename(columns={'region':'born_country_full'},inplace='True')
```

```python
bios_new.head()
```

```python
bios_new[bios_new['NOC_x']!= bios_new['born_country_full']] [['name','NOC_x','born_country_full']]
```

```python
india=bios[bios['born_country']=='IND'].copy()
russia=bios[bios['born_country']=='RUS'].copy()
```

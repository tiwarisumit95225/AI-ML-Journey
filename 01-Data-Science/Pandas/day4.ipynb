## Aggregating Data

```python
import pandas as pd
```

```python
bios=pd.read_csv('./data/bios.csv')
```

```python
bios.head()
```

```python
bios["born_city"].value_counts().tail()
```

```python
bios[bios["born_country"]=='IND']["born_region"].value_counts()
```

```python
coffee=pd.read_csv('./data/coffee.csv')
```

```python
coffee.groupby(["Coffee Type"])["Units Sold"].sum()
```

```python
coffee.groupby(["Coffee Type"]).agg({'Units Sold':'sum','price':'mean'})
```

```python
pivot=coffee.pivot(columns='Coffee Type',index='Day',values='revenue')
```

```python
pivot
```

```python
pivot.loc['Monday','Latte']
```

```python
pivot.sum()
```

```python
bios['born_date']=pd.to_datetime(bios['born_date'])
```

```python
bios.groupby(bios['born_date'].dt.year)['name'].count()
```

```python
bios.groupby(bios['born_date'].dt.year)['name'].count().reset_index().sort_values('name',ascending=False)
```

## Advanced Fuctionality

```python
coffee['yesterdays_revenue']=coffee['revenue'].shift(2)
```

```python
coffee
```

```python
coffee['pct_change']=coffee['revenue']/coffee['yesterdays_revenue']*100
```

```python
coffee
```

```python
bios['height_rank']=bios['height_cm'].rank(ascending=False)
```

```python
bios.sort_values(['height_rank'],ascending=False)
```

```python
coffee['cumulitive_revenue']=coffee['revenue'].cumsum()
```

```python
coffee
```

```python
latte=coffee[coffee['Coffee Type']=="Latte"].copy()
latte['3days']=latte['Units Sold'].rolling(3).sum()
```

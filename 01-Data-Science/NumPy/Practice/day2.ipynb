### Reorganizing arrays

```python
before=np.array([[1,2,3,4],[5,6,7,8]])
after=before.reshape((4,2))
print(after)
```

```python
# Vertically stacking vector
v1=np.array([1,2,3,4])
v2=np.array([5,6,7,8])
np.vstack((v1,v2,v1,v1))
```

```python
#horizontal stacking 
v3=np.ones((2,4))
v4=np.zeros((2,2))
np.hstack((v3,v4,v3))
```
